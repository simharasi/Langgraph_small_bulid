```python
from typing import TypedDict, List
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    value : List[int]
    name : str
    result : str
```

```python
def String_node(state : AgentState) -> AgentState:
    print(state)
    state['result'] = f"Hi there   {state['name']} ! You sum  {sum(state['value'])}"
    print(state)
    return state
```

```python
graph = StateGraph(AgentState)
graph.add_node("firstnode",String_node)
graph.set_entry_point("firstnode")
graph.set_finish_point("firstnode")
app = graph.compile()
```

```python
from IPython.display import Image, display
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
result = app.invoke({"value":[1,2,3,4,5], "name" : "Bob"})
```

```python
print(result["result"])
```

```python
from typing import TypedDict, List
from langgraph.graph import StateGraph
```

```python
class AgentState1(TypedDict):
    value : List[int]
    name : str
    result : str
    operation : str
```

```python
import math
def String_node(state : AgentState1) -> AgentState1:
    print(state)
    if(state['operation'] == '+'):
        state['result'] = f"Hi {state['name']} ! You answer {sum(state['value'])}"
    else:
        state['result'] = f"Hi there   {state['name']} ! You sum  {math.prod(state['value'])}"
    print(state)
    return state
```

```python
graph = StateGraph(AgentState1)
graph.add_node("firstnode",String_node)
graph.set_entry_point("firstnode")
graph.set_finish_point("firstnode")
app = graph.compile()
```

```python
from IPython.display import Image, display
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
result = app.invoke({"name" : "Jack Sparrow","value":[1,2,3,4],'operation':'*'})
```

```python
print(result['result'])
```
